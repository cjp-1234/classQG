```python
#使用最小二乘法对LSTAT和MEDV进行线性回归
x=bosten['LSTAT'].values #提取两项的值
y=bosten['MEDV'].values
#将x转换成矩阵 并添加一列全1的元素
x_matrix=np.c_[np.ones((x.shape[0],1)),x] 
#利用公式求系数矩阵
theta=np.linalg.inv(x_matrix.T.dot(x_matrix)).dot(x_matrix.T).dot(y)
#求y的预测值
y_prev=x_matrix.dot(theta)

#计算均方误差
mse=np.mean(y-y_prev)**2
print('均方误差:',mse)
#可视化
plt.scatter(x,y,color='blue',label='实际值')
plt.plot(x,y_prev,color='red',label='预测值')
plt.xlabel('LSTAT')
plt.ylabel('MEDV')
plt.title("线性回归:LSTAT VS MEVD")
plt.legend()
plt.show()
```

```python
print("由图像可知：总体上LSTAT越大,房价越低")  #结论
```

```python
#使用梯度下降对RM和房价进行线性回归
x = bosten['RM'].values
y = bosten['MEDV'].values

x_matrix = np.c_[np.ones((x.shape[0], 1)), x]

# 初始化参数,生成两行一列的随机矩阵
theta = np.random.randn(2, 1)

# 定义学习率和迭代次数
rate = 0.01
n = 1000

# 梯度下降
for _ in range(n):
    y_pred=x_matrix.dot(theta)  #y的预测值
    gradients = 2 / len(x_matrix) * x_matrix.T.dot(y_pred- y.reshape(-1, 1)) #梯度
    theta = theta - rate * gradients

# 计算均方误差
y_pred=x_matrix.dot(theta)
mse = np.mean((y_pred- y.reshape(-1, 1)) ** 2)

# 可视化
plt.scatter(x, y,color='blue',label='实际值')
plt.plot(x,y_pred,color='red',label='预测值' )
plt.xlabel('RM')
plt.ylabel('MEDV')
plt.title('线性回归：RM VS MEDV')
plt.legend()
plt.show()

print('线性回归参数：', theta)
print('均方误差：', mse)
```

```python
#下面对所有数据对房价进行线性回归,并进行预测
# 提取特征变量和目标变量
x= bosten.drop('MEDV',axis=1).values
y = bosten['MEDV'].values

# 数据标准化
x_mean = np.mean(x)
x_std = np.std(x)
x = (x - x_mean) / x_std

# 添加偏置项
x = np.c_[np.ones((x.shape[0], 1)), x]

rate=0.01 #学习率和迭代次数
n=1000
sample=x.shape
theta = np.zeros(sample[1])#初始化参数
#梯度下降
for _ in range(n):
    y_pred=np.dot(x,theta)
    error=y_pred-y
    gradient = (2 / sample[0]) * x.T.dot(error)#梯度
    theta =theta-rate*gradient #更新参数
    
```

```python
#导入鸢尾花数据集,进行逻辑回归
iris=pd.read_csv(r"iris.csv")
iris
```

```python
#定义sigmoid函数
def sigmoid(z):
    return 1/(1+np.exp(-z))
#定义损失函数
def cost_fun(x,y,theta):
    num=len(y)
    h=sigmoid(x.dot(theta))
    cost=-1/num*(y.T.dot(np.log(h))+(1-y).T.dot(np.log(1-h)))
    return cost
#定义梯度下降函数
def gradient_descent(x,y,theta,rate,n):
    m=len(y)
    cost_history = []   #创建一个列表记录损失情况
    for _ in range(n):
        y_pred=x.dot(theta)
        gradient=1/m*x.T.dot(y_pred-y)
        theta=theta-rate*gradient
        cost_history.append(cost_fun(x,y_pred,theta))
    return theta,cost_history     #返回系数theta和每次迭代的损失

#将目标变量转换为二分问题,setosa为1,其余为0
iris['Species']=iris['Species'].map({'setosa':1,'virginica':0,'versicolor':0})
#读取数据
x = iris[['Sepal.Length', 'Sepal.Width', 'Petal.Length', 'Petal.Width']].values
y=iris['Species'].values

#划分训练集和测试集
np.random.seed(42)
indices= np.random.permutation(len(x)) #用以打乱x
size=int(0.2*len(x))
test_indices=indices[:size]  #划分20%为测试集
train_indices=indices[size:]
#通过索引划分
x_test=x[test_indices]
x_train=x[train_indices]
y_test=y[test_indices]
y_train=y[train_indices]

#添加偏置项
x_test=np.insert(x_test,0,1,axis=1)
x_train=np.insert(x_train,0,1,axis=1)
#初始化theta
theta=np.zeros(x_train.shape[1])
#设置学习率和迭代次数,进行梯度下降
rate=0.01
n=1000
theta,cost_history=gradient_descent(x_train,y_train,theta,rate,n)

#训练完后使用测试集进行测试
h=x_test.dot(theta)
y_pred=(h>=0.5).astype(int)
#计算准确率
accuracy=np.sum(y_pred==y_test)/len(y_test)*100
print('准确率：',accuracy,'%')

# 可视化代价函数历史
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(range(n),cost_history,color='blue')
plt.xlabel('迭代次数')
plt.ylabel('代价函数值')
plt.title('代价函数历史')
#可视化测试情况
plt.subplot(1, 2, 2)
plt.scatter(range(len(y_test)),y_test,color='blue',label='真实值')
plt.scatter(range(len(y_test)),y_pred,color='red',marker='x',label='预测值')
plt.xlabel('样本索引')
plt.ylabel('类别')
plt.title('预测结果与真实值对比')
plt.legend()
plt.tight_layout()
plt.show()
```
